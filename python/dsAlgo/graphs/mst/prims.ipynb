```python
'''
We are implmenting Prim's algorithm to find the minimum spanning tree of a graph.

Definition: A minimum spanning tree is a subset of the edges of a connected, edge-weighted
undirected graph that connects all the vertices together, without any cycles and with the minimum
possible total edge weight.

FAQ:
Q1. Does Prim's algorithm work for directed graphs?
A1. No, it does not work for directed graphs. It only works for undirected graphs. The reason is
that in Prim's algorithm, we are always looking for the minimum edge weight. In a directed graph,
the edge weight from A to B can be different from the edge weight from B to A. So, we cannot use
Prim's algorithm for directed graphs.
'''
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import IPython.display as display

# display.Image(url='https://en.wikipedia.org/wiki/File:PrimAlgDemo.gif')
```

```python
adj_list = [
	{ 1: 4, 7: 8 },
	{ 0: 4, 2: 8, 7: 11 },
	{ 1: 8, 3: 7, 5: 4, 8: 2 },
	{ 2: 7, 4: 9, 5: 14 },
	{ 3: 9, 5: 10 },
	{ 2: 4, 3: 14, 4: 10, 6: 2 },
	{ 5: 2, 7: 1, 8: 6 },
	{ 0: 8, 1: 11, 6: 1, 8: 7 },
	{ 2: 2, 6: 6, 7: 7 }
]
```

```python
# Create a simple graph
G = nx.Graph()
G.add_nodes_from([i for i in range(0, len(adj_list))])
for i in range(0, len(adj_list)):
	for j in adj_list[i]:
		if adj_list[i][j] != 0:
			G.add_edge(i, j, weight=adj_list[i][j])

edge_weights = nx.get_edge_attributes(G, 'weight')
edge_colors = [1 for edge in G.edges]
node_colors = ['orange' for node in G.nodes]
```

```python
def draw():
	# Draw the graph
	pos = nx.circular_layout(G)  # You can use other layouts as well
	# nx.draw(G, pos, with_labels=True, node_size=700, node_color='orange', font_size=10, font_color='black', font_weight='bold', edge_colors=edge_colors)
	nx.draw(G, pos, with_labels=True, node_size=700, node_color=node_colors, font_size=10, font_color='black', font_weight='bold', edge_color=edge_colors, width=2, cmap=plt.cm.Blues)
	nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_weights)

	# Display the graph
	plt.show()
```

```python
draw()
```
